# Part 4 — Investment clients

The company has four different clients.

You must translate each client's preferences into measurable criteria using the available variables.

## Client A — Budget-Oriented Investor

## Client B — Premium Market Investor

## Client C — Space-Oriented Investor

## Client D — Value Investor

## Client A

```python
client_a = df.copy()

client_a["PriceScore"] = 1 - client_a["MedHouseVal"].rank(pct=True)
client_a["IncomeScore"] = client_a["MedInc"].rank(pct=True)
client_a["OccupancyScore"] = 1 - client_a["AveOccup"].rank(pct=True)

client_a["Score_A"] = (
    0.45 * client_a["PriceScore"] +
    0.40 * client_a["IncomeScore"] +
    0.15 * client_a["OccupancyScore"]
)

top_A = client_a.sort_values("Score_A", ascending=False).head(10)
display(top_A[["Latitude", "Longitude", "MedHouseVal", "MedInc", "AveOccup", "Score_A"]])
```

### Justification

- **Variables:** MedHouseVal (minimized), MedInc (maximized), and AveOccup (minimized).
- **Strategy:** Weighted percentile ranking (avoids rigid cutoffs and standardizes variable scales between 0 and 1).
- **Weights:** Price (45%) is the primary driver for a budget client; Income (40%) ensures economic viability of the area; Occupancy (15%) acts as a secondary filter against overcrowding.

## Client B

```python
client_b = df.copy()

client_b["IncomeScore"] = client_b["MedInc"].rank(pct=True)
client_b["PriceScore"] = client_b["MedHouseVal"].rank(pct=True)
client_b["RoomScore"] = client_b["AveRooms"].rank(pct=True)

client_b["Score_B"] = (
    0.40 * client_b["IncomeScore"] +
    0.40 * client_b["PriceScore"] +
    0.20 * client_b["RoomScore"]
)

top_B = client_b.sort_values("Score_B", ascending=False).head(10)
display(top_B[["Latitude", "Longitude", "MedHouseVal", "MedInc", "AveRooms", "Score_B"]])
```

### Justification

- **Variables:** MedInc (maximized), MedHouseVal (maximized), and AveRooms (maximized).
- **Strategy:** Weighted percentile ranking.
- **Weights:** Income (40%) and House Value (40%) receive equal top weighting to capture prestigious, high-capitalization areas; Room count (20%) rewards spacious residential inventory.

## Client C

```python
client_c = df.copy()

client_c["RoomScore"] = client_c["AveRooms"].rank(pct=True)
client_c["OccupancyScore"] = 1 - client_c["AveOccup"].rank(pct=True)
client_c["PriceScore"] = 1 - client_c["MedHouseVal"].rank(pct=True)

client_c["Score_C"] = (
    0.45 * client_c["RoomScore"] +
    0.35 * client_c["OccupancyScore"] +
    0.20 * client_c["PriceScore"]
)

top_C = client_c.sort_values("Score_C", ascending=False).head(10)
display(top_C[["Latitude", "Longitude", "MedHouseVal", "AveRooms", "AveOccup", "Score_C"]])
```

### Justification

- **Variables:** AveRooms (maximized), AveOccup (minimized), and MedHouseVal (minimized).
- **Strategy:** Weighted percentile ranking.
- **Weights:** Rooms (45%) prioritizes expansive properties; Occupancy (35%) ensures low neighborhood/household crowding; Price (20%) ensures reasonable asset cost.

# Part 5 — Feature engineering

Create at least **two derived variables** that may be useful for investment analysis.

```python
df["RoomsPerOccupant"] = df["AveRooms"] / df["AveOccup"]
df["BedroomRatio"] = df["AveBedrms"] / df["AveRooms"]

display(df[["AveRooms", "AveOccup", "RoomsPerOccupant", "AveBedrms", "BedroomRatio"]].head())
```

### Feature 1

Formula: RoomsPerOccupant = AveRooms / AveOccup  
Meaning: Measures the average number of physical rooms available per individual resident, serving as a density and spatial comfort indicator.  
Limitation: Aggregated at the census tract level, it does not reveal the true usable square footage of individual units.

### Feature 2

Formula: BedroomRatio = AveBedrms / AveRooms  
Meaning: Represents the proportion of total rooms dedicated to sleeping areas; lower values reflect larger living/common areas typical of higher-end single-family homes.  
Limitation: Does not account for bathrooms, functional layouts, or floor plan efficiencies.

# Part 6 — Predictive modeling

Split dataset into training and test sets and evaluate models.

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Feature matrix and target
X = df.drop(columns=["MedHouseVal"])
y = df["MedHouseVal"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 1. Baseline Model
baseline_pred = np.full(len(y_test), y_train.mean())

results = []
results.append({
    "Model": "Baseline",
    "MAE": mean_absolute_error(y_test, baseline_pred),
    "RMSE": np.sqrt(mean_squared_error(y_test, baseline_pred)),
    "R2": r2_score(y_test, baseline_pred)
})
predictions = {"Baseline": baseline_pred}
```

```python
# 2. Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
lr_pred = lr_model.predict(X_test)
predictions["Linear Regression"] = lr_pred

results.append({
    "Model": "Linear Regression",
    "MAE": mean_absolute_error(y_test, lr_pred),
    "RMSE": np.sqrt(mean_squared_error(y_test, lr_pred)),
    "R2": r2_score(y_test, lr_pred)
})
```

```python
# 3. Decision Tree Regressor
dt_model = DecisionTreeRegressor(max_depth=10, random_state=42)
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)
predictions["Decision Tree"] = dt_pred

results.append({
    "Model": "Decision Tree",
    "MAE": mean_absolute_error(y_test, dt_pred),
    "RMSE": np.sqrt(mean_squared_error(y_test, dt_pred)),
    "R2": r2_score(y_test, dt_pred)
})
```

```python
# 4. Additional Model (Random Forest Regressor)
rf_model = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)
predictions["Random Forest"] = rf_pred

results.append({
    "Model": "Random Forest",
    "MAE": mean_absolute_error(y_test, rf_pred),
    "RMSE": np.sqrt(mean_squared_error(y_test, rf_pred)),
    "R2": r2_score(y_test, rf_pred)
})
```

```python
# Compare model performance in a table
results_df = pd.DataFrame(results).sort_values("RMSE")
display(results_df)
```

### Model comparison

1. **Baseline necessity:** It provides a naïve reference point (predicting the mean target) to verify whether trained models actually learn useful patterns from the data.
2. **Best performing model:** Random Forest performs best, achieving the lowest MAE (~0.33) and RMSE (~0.49), with an R² exceeding 0.81.
3. **Predictive accuracy vs. explanation:** No. Ensemble models like Random Forest predict better by modeling complex non-linearities and interactions, but Linear Regression offers clearer, interpretable marginal coefficients.
4. **Most useful business metric:** **MAE** is the most practical metric for business communication because it directly represents the average error in property price units () without quadratic distortion.

# Part 8 — Client D: model-based value screening

```python
# Predict across entire dataset
df["PredictedValue"] = rf_model.predict(X)
df["PredictionGap"] = df["PredictedValue"] - df["MedHouseVal"]

# Apply business filters: income >= median and occupancy <= 75th percentile
income_threshold = df["MedInc"].median()
occupancy_threshold = df["AveOccup"].quantile(0.75)

candidates_D = df[
    (df["PredictionGap"] > 0) &
    (df["MedInc"] >= income_threshold) &
    (df["AveOccup"] <= occupancy_threshold)
].sort_values("PredictionGap", ascending=False)

top_D = candidates_D.head(10)
display(top_D[["Latitude", "Longitude", "MedHouseVal", "PredictedValue", "PredictionGap", "MedInc", "AveOccup"]])
```

### Interpretation

1. **Not proof of undervaluation:** A high positive gap only indicates that the model expected a higher price based on available features; it does not account for unseen negative characteristics.
2. **Alternative explanations:** Missing negative local factors (e.g., proximity to highways/landfills, high crime, low school ratings, structural deterioration, flood risk) or statistical model residuals.
3. **Additional data needed:** Rental price histories, physical inspection reports, municipal property taxes, crime indices, school district scores, and historical appreciation rates.

# Part 9 — Error analysis

```python
error_df = X_test.copy()
error_df["Observed"] = y_test.values
error_df["Predicted"] = rf_pred
error_df["Error"] = error_df["Predicted"] - error_df["Observed"]
error_df["AbsoluteError"] = error_df["Error"].abs()

print("Top 10 Largest Absolute Errors:")
display(error_df.sort_values("AbsoluteError", ascending=False).head(10))

# Split by price tiers
median_price = y_train.median()
error_df["PriceGroup"] = np.where(error_df["Observed"] < median_price, "Lower-value", "Higher-value")

print("\nError Metrics by Price Group:")
display(error_df.groupby("PriceGroup").agg(
    MAE=("AbsoluteError", "mean"),
    MeanError=("Error", "mean"),
    Count=("Error", "size")
))

# Spatial residual map
plt.figure(figsize=(9, 7))
plt.scatter(
    error_df["Longitude"],
    error_df["Latitude"],
    c=error_df["AbsoluteError"],
    cmap="Reds",
    alpha=0.6
)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Spatial Distribution of Prediction Errors (Test Set)")
plt.colorbar(label="Absolute Error (,000s)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

### Interpretation

1. **Systematic bias:** The model systematically overpredicts lower-value properties (positive mean error) and underpredicts expensive properties (negative mean error).
2. **Predictability of expensive areas:** High-value areas exhibit significantly larger MAE and error variance.
3. **Impact of target cap:** The artificial cap at ,000 creates severe truncation, forcing the model to underpredict luxury properties capped at 5.0.
4. **Practical implication:** Model predictions should only serve as an initial screening mechanism and must never replace on-the-ground appraisal.

## Bonus A — Cross-validation

```python
# Bonus A — 5-Fold Cross-Validation
from sklearn.model_selection import KFold, cross_val_score

kf = KFold(n_splits=5, shuffle=True, random_state=42)

cv_models = {
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
}

cv_results = []
for name, model in cv_models.items():
    scores = -cross_val_score(model, X, y, cv=kf, scoring="neg_root_mean_squared_error")
    cv_results.append({
        "Model": name,
        "Mean RMSE": scores.mean(),
        "Std RMSE": scores.std()
    })

display(pd.DataFrame(cv_results))
```
